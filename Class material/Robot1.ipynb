```python
import time


def move_servo(servo, angle):
    print(f"Servo {servo} moving to {angle} degrees")
    time.sleep(0.5)


def open_gripper():
    print("Gripper OPEN")
    time.sleep(0.5)


def close_gripper():
    print("Gripper CLOSE")
    time.sleep(0.5)


print("=== Robot Pick-and-Place Test ===")

# Starting position
move_servo(1, 90)
move_servo(2, 90)
move_servo(3, 90)
move_servo(4, 90)

# Pick up an object
print("\nPicking object...")
open_gripper()
move_servo(2, 120)
close_gripper()

# Move object
print("\nMoving object...")
move_servo(1, 150)
move_servo(2, 80)

# Place object
print("\nPlacing object...")
open_gripper()

# Return home
print("\nReturning home...")
move_servo(1, 90)
move_servo(2, 90)

print("\nTask complete!")
```

```python
import time


class RobotArm:
    def __init__(self, name):
        self.name = name
        self.joint_angles = [90, 90, 90, 90]
        self.gripper_open = True

    def move_joint(self, joint, angle):
        # Check whether the joint number is valid
        if joint < 1 or joint > 4:
            print("Error: Joint must be between 1 and 4.")
            return

        # Check whether the angle is safe
        if angle < 0 or angle > 180:
            print("Error: Angle must be between 0 and 180 degrees.")
            return

        self.joint_angles[joint - 1] = angle

        print(f"Joint {joint} moving to {angle} degrees")
        time.sleep(0.5)

    def open_gripper(self):
        if self.gripper_open:
            print("Gripper is already open.")
        else:
            self.gripper_open = True
            print("Gripper OPEN")

        time.sleep(0.5)

    def close_gripper(self):
        if not self.gripper_open:
            print("Gripper is already closed.")
        else:
            self.gripper_open = False
            print("Gripper CLOSE")

        time.sleep(0.5)

    def move_home(self):
        print("\nReturning to home position...")

        for joint in range(1, 5):
            self.move_joint(joint, 90)

    def show_status(self):
        print("\n--- Robot Status ---")

        for i, angle in enumerate(self.joint_angles):
            print(f"Joint {i + 1}: {angle} degrees")

        if self.gripper_open:
            print("Gripper: OPEN")
        else:
            print("Gripper: CLOSED")


# --------------------------------
# Main Program
# --------------------------------

robot = RobotArm("RoboArm-M2-S")

print(f"Starting {robot.name}")

robot.move_home()

# Objects waiting for the robot
objects = ["red block", "blue block", "green block"]

for object_name in objects:

    print(f"\n===== Processing {object_name} =====")

    # Pick position
    robot.move_joint(1, 60)
    robot.move_joint(2, 120)

    robot.open_gripper()
    robot.close_gripper()

    # Decide where to place the object
    if object_name == "red block":
        target_angle = 30

    elif object_name == "blue block":
        target_angle = 90

    else:
        target_angle = 150

    print(f"{object_name} will be placed at {target_angle} degrees.")

    # Move to target
    robot.move_joint(1, target_angle)
    robot.move_joint(2, 70)

    # Release object
    robot.open_gripper()

    # Return to ready position
    robot.move_joint(2, 90)


robot.move_home()
robot.show_status()

print("\nAll objects have been processed!")
```

```python
# Interactive Robot Arm Control Panel
# Run the RobotArm class cell before running this cell.
import ipywidgets as widgets
from IPython.display import display

control_robot = RobotArm("RoboArm-M2-S")

title = widgets.HTML("""
<div style='padding:18px;border-radius:14px;background:#1e3a8a;color:white'>
  <div style='font-size:12px;letter-spacing:2px'>ROBOTICS LAB</div>
  <div style='font-size:28px;font-weight:bold'>RoboArm Control Center</div>
  <div>Position the joints, operate the gripper, and load saved poses.</div>
</div>
""")

sliders = []
for joint in range(1, 5):
    slider = widgets.IntSlider(
        value=90, min=0, max=180, description="Joint " + str(joint),
        continuous_update=False, layout=widgets.Layout(width="95%")
    )
    sliders.append(slider)

status = widgets.HTML()
activity = widgets.Output(layout=widgets.Layout(
    height="140px", overflow_y="auto", border="1px solid #bfdbfe", padding="8px"
))
apply_button = widgets.Button(description="Apply Position", button_style="primary", icon="play")
home_button = widgets.Button(description="Move Home", icon="home")
open_button = widgets.Button(description="Open Gripper", button_style="success", icon="unlock")
close_button = widgets.Button(description="Close Gripper", button_style="warning", icon="lock")
pose_menu = widgets.Dropdown(
    options=["Ready", "Pick", "Place Left", "Place Center", "Place Right"],
    description="Saved pose"
)
load_button = widgets.Button(description="Load Pose", icon="download")

poses = {
    "Ready": [90, 90, 90, 90],
    "Pick": [60, 120, 90, 90],
    "Place Left": [30, 70, 90, 90],
    "Place Center": [90, 70, 90, 90],
    "Place Right": [150, 70, 90, 90]
}

def update_status():
    joint_text = ""
    for number in range(4):
        joint_text += "<b>Joint " + str(number + 1) + ":</b> " + str(control_robot.joint_angles[number]) + "° &nbsp; "
    gripper_text = "OPEN" if control_robot.gripper_open else "CLOSED"
    status.value = "<div style='padding:14px;background:#eff6ff;border-radius:10px'>" + joint_text + "<br><b>Gripper:</b> " + gripper_text + "</div>"

def add_log(message):
    with activity:
        print("•", message)

def apply_position(button):
    for number in range(4):
        control_robot.move_joint(number + 1, sliders[number].value)
    update_status()
    add_log("New joint position applied")

def move_home(button):
    for slider in sliders:
        slider.value = 90
    control_robot.move_home()
    update_status()
    add_log("Robot returned home")

def change_gripper(open_gripper):
    if open_gripper:
        control_robot.open_gripper()
    else:
        control_robot.close_gripper()
    update_status()
    add_log("Gripper updated")

def load_pose(button):
    angles = poses[pose_menu.value]
    for number in range(4):
        sliders[number].value = angles[number]
    add_log(pose_menu.value + " pose loaded")

apply_button.on_click(apply_position)
home_button.on_click(move_home)
open_button.on_click(lambda button: change_gripper(True))
close_button.on_click(lambda button: change_gripper(False))
load_button.on_click(load_pose)

update_status()
left_panel = widgets.VBox(sliders + [widgets.HBox([apply_button, home_button]), widgets.HBox([open_button, close_button]), widgets.HBox([pose_menu, load_button])])
right_panel = widgets.VBox([widgets.HTML("<h3>Live Robot Status</h3>"), status, widgets.HTML("<h3>Activity</h3>"), activity])
dashboard = widgets.VBox([title, widgets.HBox([left_panel, right_panel], layout=widgets.Layout(gap="25px", padding="14px"))])

add_log("Control panel ready")
display(dashboard)
```

```python
# Simple Robot Arm User Interface
# This version uses only basic Python: print, input, lists, loops, and if statements.

joint_angles = [90, 90, 90, 90]
gripper_open = True
running = True

while running:
    # Show the important robot data
    print("\n==============================")
    print("   ROBOARM CONTROL PANEL")
    print("==============================")
    print("Joint 1 angle:", joint_angles[0], "degrees")
    print("Joint 2 angle:", joint_angles[1], "degrees")
    print("Joint 3 angle:", joint_angles[2], "degrees")
    print("Joint 4 angle:", joint_angles[3], "degrees")

    if gripper_open:
        print("Gripper: OPEN")
    else:
        print("Gripper: CLOSED")

    # Show the choices
    print("\n1 - Move a joint")
    print("2 - Open gripper")
    print("3 - Close gripper")
    print("4 - Move robot home")
    print("5 - Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        joint = int(input("Enter joint number (1 to 4): "))
        angle = int(input("Enter angle (0 to 180): "))

        if joint >= 1 and joint <= 4 and angle >= 0 and angle <= 180:
            joint_angles[joint - 1] = angle
            print("Joint", joint, "moved to", angle, "degrees.")
        else:
            print("Invalid joint number or angle.")

    elif choice == "2":
        gripper_open = True
        print("Gripper opened.")

    elif choice == "3":
        gripper_open = False
        print("Gripper closed.")

    elif choice == "4":
        joint_angles = [90, 90, 90, 90]
        print("Robot moved to the home position.")

    elif choice == "5":
        running = False
        print("Robot control program ended.")

    else:
        print("Please choose a number from 1 to 5.")
```
