```python
import time


def move_servo(servo, angle):
    print(f"Servo {servo} moving to {angle} degrees")
    time.sleep(0.5)


def open_gripper():
    print("Gripper OPEN")
    time.sleep(0.5)


def close_gripper():
    print("Gripper CLOSE")
    time.sleep(0.5)


print("=== Robot Pick-and-Place Test ===")

# Starting position
move_servo(1, 90)
move_servo(2, 90)
move_servo(3, 90)
move_servo(4, 90)

# Pick up an object
print("\nPicking object...")
open_gripper()
move_servo(2, 120)
close_gripper()

# Move object
print("\nMoving object...")
move_servo(1, 150)
move_servo(2, 80)

# Place object
print("\nPlacing object...")
open_gripper()

# Return home
print("\nReturning home...")
move_servo(1, 90)
move_servo(2, 90)

print("\nTask complete!")
```

```python
import time


class RobotArm:
    def __init__(self, name):
        self.name = name
        self.joint_angles = [90, 90, 90, 90]
        self.gripper_open = True

    def move_joint(self, joint, angle):
        # Check whether the joint number is valid
        if joint < 1 or joint > 4:
            print("Error: Joint must be between 1 and 4.")
            return

        # Check whether the angle is safe
        if angle < 0 or angle > 180:
            print("Error: Angle must be between 0 and 180 degrees.")
            return

        self.joint_angles[joint - 1] = angle

        print(f"Joint {joint} moving to {angle} degrees")
        time.sleep(0.5)

    def open_gripper(self):
        if self.gripper_open:
            print("Gripper is already open.")
        else:
            self.gripper_open = True
            print("Gripper OPEN")

        time.sleep(0.5)

    def close_gripper(self):
        if not self.gripper_open:
            print("Gripper is already closed.")
        else:
            self.gripper_open = False
            print("Gripper CLOSE")

        time.sleep(0.5)

    def move_home(self):
        print("\nReturning to home position...")

        for joint in range(1, 5):
            self.move_joint(joint, 90)

    def show_status(self):
        print("\n--- Robot Status ---")

        for i, angle in enumerate(self.joint_angles):
            print(f"Joint {i + 1}: {angle} degrees")

        if self.gripper_open:
            print("Gripper: OPEN")
        else:
            print("Gripper: CLOSED")


# --------------------------------
# Main Program
# --------------------------------

robot = RobotArm("RoboArm-M2-S")

print(f"Starting {robot.name}")

robot.move_home()

# Objects waiting for the robot
objects = ["red block", "blue block", "green block"]

for object_name in objects:

    print(f"\n===== Processing {object_name} =====")

    # Pick position
    robot.move_joint(1, 60)
    robot.move_joint(2, 120)

    robot.open_gripper()
    robot.close_gripper()

    # Decide where to place the object
    if object_name == "red block":
        target_angle = 30

    elif object_name == "blue block":
        target_angle = 90

    else:
        target_angle = 150

    print(f"{object_name} will be placed at {target_angle} degrees.")

    # Move to target
    robot.move_joint(1, target_angle)
    robot.move_joint(2, 70)

    # Release object
    robot.open_gripper()

    # Return to ready position
    robot.move_joint(2, 90)


robot.move_home()
robot.show_status()

print("\nAll objects have been processed!")
```

```python
# Dictionary example for teaching students
# A dictionary stores data as key-value pairs.

robot = {
    "name": "RoboArm-M2-S",
    "speed": 3,
    "battery": 85,
    "gripper_open": True,
    "positions": [90, 120, 60],
}

print("Robot dictionary:")
print(robot)

# Access values by key
print("\nRobot name:", robot["name"])
print("Battery level:", robot["battery"], "%")

# Add a new key-value pair
robot["arm_count"] = 4
print("\nAfter adding a new key:")
print(robot)

# Update an existing value
robot["battery"] = 90
print("\nBattery updated:", robot["battery"], "%")

# Loop through dictionary items
print("\nDictionary items:")
for key, value in robot.items():
    print(f"{key}: {value}")

# Example: dictionary of robot tasks
robot_tasks = {
    "move_home": [90, 90, 90, 90],
    "pick_object": [60, 120],
    "place_object": [150, 70],
}

print("\nRobot tasks:")
print(robot_tasks)
print("Move home sequence:", robot_tasks["move_home"])
```
